```python
import numpy as np

x = np.linspace(0, 2 * np.pi, 10)
y = np.sin(x)

print("x =", x)
print("y =", y)
```

```python
import matplotlib.pyplot as plt

plt.scatter(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.title('Sine Function')
plt.grid(False)
plt.show()
```

```python
x = np.linspace(0, 2 * np.pi)
y = np.sin(x)

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Sine Function")
plt.grid(True)
plt.show()
```
